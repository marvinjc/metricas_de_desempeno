```python
#pip install -q "kagglehub[pandas-datasets]" openpyxl
import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python

# Nombre del archivo dentro del conjunto de datos de Kaggle
file_path = "Telco_customer_churn.xlsx"
```

```python
# Cargar los datos en un DataFrame
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "yeanzc/telco-customer-churn-ibm-dataset",
    file_path,
    pandas_kwargs={"engine": "openpyxl"}
)

# Mostrar las primeras cinco filas
display(df.head())
```

```python
print("Dimensiones del dataset:", df.shape)
print("\nColumnas:")
print(df.columns.tolist())
```

```python
target = "Churn Label"
conteo = df[target].value_counts()
porcentaje = df[target].value_counts(normalize=True) * 100
```

```python
print("Cantidad por clase:")
print(conteo)
print("\nPorcentaje por clase:")
print(porcentaje.round(2))
```
